```python
Xdf = df[['score', 'attitude']]
Ydf = df['status']

x = np.array(Xdf).reshape(-1, 2).astype('float64')
y = np.array(Ydf).reshape(-1, 1).astype('float64')

print(x, x.shape)
print(y, y.shape)

print(50* "=")
Pre = lambda za: 1 / (1 + np.exp(-za))
m, n = x.shape
print(f"m: {m}\nn: {n}")
lr = 0.0001

weight = np.zeros(n)
weight = np.array(weight).reshape(n, 1)
print(f"weight: {weight}\nweight shape: {weight.shape}")
bias = 0.0

print("z = w * x + b")
z = np.dot(x, weight) + bias
print(f"z: {z}\nz shape{z.shape}")

y_pred = Pre(z)
print(f"ypred:\n{y_pred}\nypred: {y_pred.shape}")
y_pred = np.array(y_pred).reshape(y_pred.shape[0], 1)
print("e = y_pred - y")
error = y_pred - y
print(f"error:\n{error}\nerror shape: {error.shape}")

db = np.sum(error) / m
print(f"db:\n{db}\ndb shape: {db.shape}")

dw = np.sum(np.dot(x.T, error) / m)
print(f"dw:\n{dw}\ndw shape: {dw.shape}")

weight = np.array(weight).reshape(n, 1)
weight -= np.dot(dw, lr)
print(f"weight:\n{weight}\nshape: {weight.shape}")

bias = bias - lr * db
print(f"bias: {bias}")
print(50* "=")
print(f"weight:\n{weight}\nweight shape: {weight.shape}\nbias: {bias}")
```

```python
class LogisticRegression:
    def __init__(self):
        self.bias = 0.0
        self.weight = 0.0

    def train(self, x, y, epoch, lr):
        m, n = x.shape[0]
        self.n = x.shape[1]
        
        self.weight = np.zeros(n)
        self.weight = np.array(self.weight).reshape(n, 1)
        self.P = lambda za: 1 / (1 + np.exp(-za))
        
        for i in range(epoch):
            z = np.dot(x, self.weight) + self.bias
            y_pred = self.P(z)

            ypred = np.array(y_pred).reshape(m, 1)
            error = ypred - y
            error = error.reshape(1, -1)
        
            dw = np.dot(error, x) / m
            db = np.sum(error) / m
            
            self.weight = self.weight - (lr * dw.T)
            self.bias -= (db * lr)
            
    def predict(self, x):
        Z = np.dot(x, self.weight) + self.bias
        P = 1 / (1 + np.exp(-Z))
        return 1 if P >= 0.5 else 0

print('creating model done!')
```

```python
%%cython -a
import cython
import numpy as np
cimport numpy as np

@cython.boundscheck(False)
@cython.wraparound(False)
@cython.cdivision(True)

cdef class LogisticRegressionC:
    cdef public double bias
    cdef public double[:] weight
    cdef int n

    def __init__(self):
        self.bias = 0.0

    property weight_arr:
        def __get__(self):
            return np.asarray(self.weight)

    cpdef void fit(self, double[:, :] x, double[:, :] y, int epoch, double lr):
        cdef int m = x.shape[0]
        self.n = x.shape[1]
        self.weight = np.zeros(self.n, dtype=np.float64)
        
        cdef int i, j, k
        cdef double[:] y_pred = np.zeros(m, dtype=np.float64)
        cdef double z, error_i, db, dw_j

        for i in range(epoch):
            db = 0.0
            dw = np.zeros(self.n, dtype=np.float64)
            
            for j in range(m):
                z = 0.0
                for k in range(self.n):
                    z += x[j, k] * self.weight[k]
                z += self.bias
                
                y_pred[j] = 1.0 / (1.0 + np.exp(-z))
                error_i = y_pred[j] - y[j, 0]
                
                db += error_i
                for k in range(self.n):
                    dw[k] += error_i * x[j, k]
            
            self.bias -= (lr * db / m)
            for k in range(self.n):
                self.weight[k] -= (lr * dw[k] / m)

    cpdef long long[:] predict(self, long long[:, :] x):
        cdef int m = x.shape[0]  
        cdef int n = x.shape[1]
        cdef long long[:] pred = np.zeros(m, dtype=np.int64)
        
        cdef int i, j
        cdef double z, p
    
        for i in range(m):
            z = 0.0
            for j in range(n):
                z += x[i, j] * self.weight[j]
            z += self.bias
            
            p = 1.0 / (1.0 + np.exp(-z))
            pred[i] = 1 if p >= 0.5 else 0
            
        return np.asarray(pred)
```

```python
%%cython -a
import cython
import numpy as np
cimport numpy as np

@cython.boundscheck(False)
@cython.wraparound(False)
@cython.cdivision(True)

cdef class CLinearROLS:
    cdef public double[:] weight
    cdef public double bias
    
    def __init__(self):
        self.bias = 0.0

    property weight_arr:
        def __get__(self):
            return np.asarray(self.weight)
    
    cpdef void fit(self, double[:, :] X, double[:, :] Y):
        cdef double sigmaX=0.0, sigmaY=0.0, sigmaXS=0.0, sigmaXY=0.0
    
        cdef int i, n, m
        n = X.shape[0]
        m = X.shape[1]

        self.weight = np.zeros(m, dtype=np.float64)

        for i in range(n):
            sigmaY += Y[i, 0]
            for j in range(m):
                sigmaXS += X[i, j] ** 2
                sigmaX += X[i, j]
                sigmaXY += X[i, j] * Y[i, 0]

        cdef double denom = (n * sigmaXS) - (sigmaX ** 2)
        for i in range(m):
            if denom != 0:
                self.weight[j] = ((n * sigmaXY) - (sigmaX * sigmaY)) / denom
        
        self.bias = (sigmaY - (self.weight[0] * sigmaX)) / n 

    cpdef double[:] predict(self, long long[:, :] X):
        cdef int i, j, n, m
        n = X.shape[0]
        m = X.shape[1]
        cdef double[:] yPred = np.zeros(n, dtype=np.float64)

        for i in range(n):
            yPred[i] = self.bias
            for j in range(m):
                yPred[i] += self.weight[j] * X[i, j]
        return yPred
```
